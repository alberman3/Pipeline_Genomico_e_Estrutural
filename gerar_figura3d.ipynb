```python
import py3Dmol

# PDB ID para a estrutura de GAPDH em Gallus gallus
pdb_id = "1u8f"

# Instancia o visualizador 3D
view = py3Dmol.view(query=f"pdb:{pdb_id}", width=800, height=600)

# Define o estilo de cartoon (fita tridimensional)
view.setStyle({'cartoon': {'color': 'spectrum'}})

# Destaque de cores para as estruturas secundárias:
# Hélices Alfa -> Vermelho / Rosa
# Folhas Beta -> Amarelo / Azul
view.setStyle({'ss': 'h'}, {'cartoon': {'color': 'red'}})     # Hélices Alfa
view.setStyle({'ss': 's'}, {'cartoon': {'color': 'yellow'}})  # Folhas Beta

# Centraliza e renderiza
view.zoomTo()
view.show()
```
